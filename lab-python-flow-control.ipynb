```python
#1. Define a list called products that contains the following items: "t-shirt", "mug", "hat", "book", "keychain".
```

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
#2. Create an empty dictionary called inventory.
```

```python
inventory = {}
```

```python
#3. Ask the user to input the quantity of each product available in the inventory. Use the product names from the products list as keys in the inventory dictionary and assign the respective quantities as values.
```

```python
for item in products:
    quantity = int(input(f"Enter {item} quantity: "))
    inventory[item] = quantity

print("Inventory:", inventory)
```

```python
#4. Create an empty set calledcustomer_orders.
```

```python
customer_orders = set()
```

```python
#5. Ask the user to input the name of three products that a customer wants to order (from those in the products list, meaning three products out of "t-shirt", "mug", "hat", "book" or "keychain". Add each product name to the customer_orders set.
```

```python
customer_orders = set()

add_product = "yes"

while add_product == "yes":
    item = input("What would you like to order: t-shirt, mug, hat, book or keychain? ").lower()

    if item in products:
        customer_orders.add(item)
    else:
        print("Invalid product or wrongly written. Please choose one from the list.")

    add_product = input("Do you want to add another product? (yes/no) ").lower()

print(customer_orders)
```

```python
total_products_ordered = len(customer_orders)
percentage_products_ordered = (total_products_ordered / sum(inventory.values())) * 100

print(total_products_ordered)
print(percentage_products_ordered)
```
